```python
# 1. Install dependencies (added 'accelerate' which modern models require)
!pip install -q transformers torch gradio accelerate

import torch
import gradio as gr
from transformers import pipeline

# Auto-detect Colab T4 GPU
device = 0 if torch.cuda.is_available() else -1

print("Loading pre-trained transformer model (text-generation)...")
# Using the text-generation pipeline and a compatible chat model
qa_pipeline = pipeline(
    "text-generation",
    model="TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    device=device
)

def generate_answer(question, history):
    # Format the prompt using TinyLlama's required chat template
    prompt = f"<|system|>\nYou are a helpful AI assistant that answers questions clearly.</s>\n<|user|>\n{question}</s>\n<|assistant|>\n"

    try:
        result = qa_pipeline(
            prompt,
            max_new_tokens=200,
            do_sample=True,
            temperature=0.7,
            return_full_text=False # Ensures it only returns the answer, not your original question
        )
        return result[0]['generated_text'].strip()
    except Exception as e:
        return f"An error occurred: {str(e)}"

demo = gr.ChatInterface(
    fn=generate_answer,
    title="🤖 Transformer Q&A System",
    description="Text Generation using a Pre-trained Transformer Model (TinyLlama).",
    textbox=gr.Textbox(placeholder="Ask a question...", container=True)
)

# Launch with share=True to generate the public Gradio link
demo.launch(debug=True, share=True)
```
